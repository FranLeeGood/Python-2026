```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
lenguajes = ['Python', 'C++', 'Java', 'JavaScript', 'C#']
print(lenguajes)
preferencia = [45, 20, 15, 12, 8]
print(preferencia)
```

```python
plt.figure(figsize= (8,5))
plt.title('Lenguajes de programacion mas usados')
plt.bar(lenguajes, preferencia, color = (['blue', 'black', 'yellow', 'gold', 'green']))
plt.xlabel('Preferencias', fontsize = 12)
plt.ylabel('lenguaje', fontsize = 12)
plt.show()
```
