```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
datos = pd.read_csv("data/Pokemon.csv", index_col="Number")
datos.head()
```

```python
datos.info()
datos.describe()
```

```python
duplicados = datos.duplicated().sum()
print("Cantidad de filas duplicadas existentes:", duplicados)

datos_sin_duplicados = datos.drop_duplicates()
```

```python
nulos_columnas = datos_sin_duplicados.isnull().sum()
print("Valores nulos por columnas:")
print(nulos_columnas)

nulos_filas = datos_sin_duplicados.isnull().sum(axis=1)
print("\nValores nulos por filas (total):", nulos_filas.sum())

datos_limpio = datos_sin_duplicados.dropna()
```

```python
def filtrar_por_tipo_fuego(dataframe):
    fuego = dataframe[dataframe["Type 1"].str.lower() == "fire"]
    return fuego

datos_fuego = filtrar_por_tipo_fuego(datos_limpio)
datos_fuego.head(20)
```

```python
promedio_defensa_fuego = datos_fuego["Def"].mean()
print("El promedio de defensa de los pokemons tipo fuego es:", promedio_defensa_fuego)
```

```python
# 1. Gráfico de Barra: Top 10 Pokémon tipo Fuego con más Ataque
top_10_ataque = datos_fuego.nlargest(10, "Att")

plt.figure(figsize=(10, 5))
plt.bar(top_10_ataque["Name"], top_10_ataque["Att"], color="orangered")
plt.title("Top 10 Pokemons Tipo Fuego por su Nivel de Ataque")
plt.xlabel("Nombre del Pokemon")
plt.ylabel("Puntos de Ataque")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 2. Gráfico de Línea: Cantidad de Pokémon tipo Fuego por Generación
fuego_generacion = datos_fuego["Generation"].value_counts().sort_index()

plt.figure(figsize=(10, 5))
plt.plot(fuego_generacion.index, fuego_generacion.values, marker="o", color="red", linewidth=2)
plt.title("Cantidad de Pokemons Tipo Fuego por Generacion")
plt.xlabel("Generacion")
plt.ylabel("Cantidad de Pokemons")

for x, y in zip(fuego_generacion.index, fuego_generacion.values):
    plt.text(x, y + 0.5, str(y), ha="center", va="bottom")

plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```
