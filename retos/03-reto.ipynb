```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def preparar_datos():
    horas_de_simulacion = np.arange(1,11)      # Array = [1,2,3,4,5,6,7,8]
    uso_de_cpu_registrado = [45, 58, 52, 70, 85, 92, 78, 64, 61, 50]

    data = {
        'Hora' : horas_de_simulacion,
        '% Uso de CPU' : uso_de_cpu_registrado
    }
    # Transformando un diccionario a data frame
    tata = pd.DataFrame(data)
    return tata
```

```python
def graficar_rendimiento(dataFrame):
    fig, ax = plt.subplots()            

    ax.plot(dataFrame['Hora'], dataFrame['% Uso de CPU'], marker = 'o', color = 'green', linestyle = '-')
    ax.set_title('Tasas de uso de memoria CPU (en %) de un servidor de cómputo durante las primeras 10 horas')
    ax.set_xlabel('Horas de simulación')
    ax.set_ylabel('% Uso de CPU')

    ax.grid(True)
    fig.savefig('gradico01', dpi = 300)
```

```python
mostrar_grafico = preparar_datos()
graficar_rendimiento(mostrar_grafico)
plt.show()
```
